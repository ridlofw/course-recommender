## 2) Content-based Recommender — User Profile (Genres)

```python
from content_user_profile import recommend
uid = int(train['user_id'].sample(1, random_state=42).iloc[0])
recs = recommend(uid, train, data.courses, top_k=10)
titles = data.courses.set_index('course_id')['title'].to_dict()
uid, [(cid, titles.get(cid,'')) for cid in recs]
```

## 3) Content-based Recommender — Course Similarity (TF-IDF)

```python
from content_similarity import similar_courses
seed = int(data.courses['course_id'].sample(1, random_state=1).iloc[0])
recs = similar_courses(data.courses, seed_course_id=seed, top_k=10)
(seed, titles.get(seed,'')), [(cid, titles.get(cid,'')) for cid in recs]
```
